```python
# Import necessary libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from sklearn.metrics import classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

# Set style and seeds
plt.style.use('seaborn-v0_8')
np.random.seed(42)
tf.random.set_seed(42)

print("Libraries imported successfully!")

# Load CIFAR-10 and trained model
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

class_names = ['Airplane', 'Automobile', 'Bird', 'Cat', 'Deer', 
               'Dog', 'Frog', 'Horse', 'Ship', 'Truck']

print("Data loaded successfully!")

# Create and train a simple model for visualization
def create_simple_cnn():
    """Create a simple CNN for visualization purposes"""
    model = keras.Sequential([
        layers.Conv2D(32, 3, activation='relu', input_shape=(32, 32, 3), name='conv1'),
        layers.MaxPooling2D(2, name='pool1'),
        layers.Conv2D(64, 3, activation='relu', name='conv2'),
        layers.MaxPooling2D(2, name='pool2'),
        layers.Conv2D(128, 3, activation='relu', name='conv3'),
        layers.Flatten(name='flatten'),
        layers.Dense(512, activation='relu', name='dense1'),
        layers.Dense(10, activation='softmax', name='output')
    ])
    
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# Create and train model
vis_model = create_simple_cnn()
print("Model created for visualization!")

# Function to visualize filters
def visualize_filters(model, layer_name, n_filters=8):
    """Visualize learned filters from a convolutional layer"""
    
    # Get the layer
    layer = model.get_layer(layer_name)
    weights = layer.get_weights()[0]  # Get filter weights
    
    # Normalize weights for visualization
    weights = (weights - weights.min()) / (weights.max() - weights.min())
    
    # Plot filters
    fig, axes = plt.subplots(2, n_filters//2, figsize=(12, 6))
    axes = axes.flatten()
    
    for i in range(min(n_filters, weights.shape[-1])):
        if weights.shape[-2] == 3:  # RGB filters
            axes[i].imshow(weights[:, :, :, i])
        else:  # Grayscale filters
            axes[i].imshow(weights[:, :, 0, i], cmap='viridis')
        axes[i].set_title(f'Filter {i+1}')
        axes[i].axis('off')
    
    plt.suptitle(f'Learned Filters from {layer_name}', fontsize=16)
    plt.tight_layout()
    plt.show()

print("Visualization functions ready!")
```
